```python

!pip install -q pytest

import pytest
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

URL = "https://www.saucedemo.com/"
USERNAME = "standard_user"
PASSWORD = "secret_sauce"


def login(driver, username=USERNAME, password=PASSWORD):
    driver.get(URL)
    driver.find_element(By.ID, "user-name").send_keys(username)
    driver.find_element(By.ID, "password").send_keys(password)
    driver.find_element(By.ID, "login-button").click()


# TC01: Đăng nhập thành công
def test_TC01_login_success():
    login(driver)

    assert "/inventory.html" in driver.current_url
    assert driver.find_element(
        By.CLASS_NAME, "inventory_list"
    ).is_displayed()


# TC02: Đăng nhập sai mật khẩu
def test_TC02_login_wrong_password():
    login(driver, USERNAME, "wrong_password")

    error = WebDriverWait(driver, 5).until(
        EC.visibility_of_element_located(
            (By.CSS_SELECTOR, "[data-test='error']")
        )
    )
    assert "Username and password do not match" in error.text


# TC03: Hiển thị danh sách sản phẩm
def test_TC03_display_products():
    login(driver)

    products = driver.find_elements(
        By.CLASS_NAME, "inventory_item"
    )
    assert len(products) > 0


# TC04: Thêm sản phẩm vào giỏ hàng
def test_TC04_add_to_cart():
    login(driver)

    driver.find_element(
        By.ID, "add-to-cart-sauce-labs-backpack"
    ).click()

    badge = driver.find_element(
        By.CLASS_NAME, "shopping_cart_badge"
    )
    assert badge.text == "1"


# TC05: Đăng xuất
def test_TC05_logout():
    login(driver)

    driver.find_element(
        By.ID, "react-burger-menu-btn"
    ).click()

    logout_link = WebDriverWait(driver, 5).until(
        EC.element_to_be_clickable(
            (By.ID, "logout_sidebar_link")
        )
    )
    logout_link.click()

    assert driver.find_element(
        By.ID, "login-button"
    ).is_displayed()


# Chạy lần lượt 5 test case
test_cases = [
    test_TC01_login_success,
    test_TC02_login_wrong_password,
    test_TC03_display_products,
    test_TC04_add_to_cart,
    test_TC05_logout
]

results = []

for test in test_cases:
    try:
        test()
        results.append((test.__name__, "PASSED"))
        print("PASSED:", test.__name__)
    except Exception as e:
        results.append((test.__name__, "FAILED"))
        print("FAILED:", test.__name__, "-", str(e)[:250])

print("\n===== TỔNG KẾT KIỂM THỬ =====")
print("Tổng số:", len(results))
print("Thành công:", sum(r[1] == "PASSED" for r in results))
print("Thất bại:", sum(r[1] == "FAILED" for r in results))
```

```python

import glob
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

URL = "https://www.saucedemo.com/"
USERNAME = "standard_user"
PASSWORD = "secret_sauce"

chrome_paths = glob.glob(
    "/root/.cache/ms-playwright/chromium-*/chrome-linux64/chrome"
)
CHROME_PATH = chrome_paths[0]

def create_driver():
    options = Options()
    options.binary_location = CHROME_PATH
    options.add_argument("--headless")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920,1080")
    return webdriver.Chrome(options=options)

def login(driver, username=USERNAME, password=PASSWORD):
    driver.get(URL)
    wait = WebDriverWait(driver, 15)
    wait.until(EC.visibility_of_element_located((By.ID, "user-name"))).send_keys(username)
    driver.find_element(By.ID, "password").send_keys(password)
    driver.find_element(By.ID, "login-button").click()
    return wait

def run_test(name, test_function):
    browser = None
    try:
        browser = create_driver()
        test_function(browser)
        print("PASSED:", name)
        return "PASSED"
    except Exception as e:
        print("FAILED:", name, "-", str(e).splitlines()[0])
        return "FAILED"
    finally:
        if browser:
            browser.quit()

# TC01: Đăng nhập thành công
def tc01(browser):
    login(browser)
    WebDriverWait(browser, 15).until(
        EC.visibility_of_element_located((By.CLASS_NAME, "inventory_list"))
    )
    assert "/inventory.html" in browser.current_url

# TC02: Đăng nhập sai mật khẩu
def tc02(browser):
    login(browser, USERNAME, "wrong_password")
    error = WebDriverWait(browser, 15).until(
        EC.visibility_of_element_located((By.CSS_SELECTOR, "[data-test='error']"))
    )
    assert error.is_displayed()
    assert "do not match" in error.text.lower()

# TC03: Hiển thị danh sách sản phẩm
def tc03(browser):
    login(browser)
    WebDriverWait(browser, 15).until(
        EC.visibility_of_element_located((By.CLASS_NAME, "inventory_list"))
    )
    products = browser.find_elements(By.CLASS_NAME, "inventory_item")
    assert len(products) > 0, "Không tìm thấy sản phẩm"

# TC04: Thêm balo vào giỏ hàng
def tc04(browser):
    login(browser)
    wait = WebDriverWait(browser, 15)
    wait.until(EC.visibility_of_element_located((By.CLASS_NAME, "inventory_list")))
    wait.until(EC.element_to_be_clickable(
        (By.ID, "add-to-cart-sauce-labs-backpack")
    )).click()
    badge = wait.until(EC.visibility_of_element_located(
        (By.CLASS_NAME, "shopping_cart_badge")
    ))
    assert badge.text == "1"

# TC05: Đăng xuất
def tc05(browser):
    login(browser)
    wait = WebDriverWait(browser, 15)
    wait.until(EC.visibility_of_element_located((By.CLASS_NAME, "inventory_list")))
    wait.until(EC.element_to_be_clickable(
        (By.ID, "react-burger-menu-btn")
    )).click()
    wait.until(EC.element_to_be_clickable(
        (By.ID, "logout_sidebar_link")
    )).click()
    wait.until(EC.visibility_of_element_located((By.ID, "login-button")))
    assert browser.current_url.rstrip("/") == URL.rstrip("/")

tests = [
    ("TC01 - Login success", tc01),
    ("TC02 - Wrong password", tc02),
    ("TC03 - Display products", tc03),
    ("TC04 - Add to cart", tc04),
    ("TC05 - Logout", tc05),
]

results = [(name, run_test(name, fn)) for name, fn in tests]

print("\n===== TEST SUMMARY =====")
print("Total:", len(results))
print("Passed:", sum(status == "PASSED" for _, status in results))
print("Failed:", sum(status == "FAILED" for _, status in results))
```

```python

browser = create_driver()

try:
    wait = WebDriverWait(browser, 15)
    login(browser)

    wait.until(
        EC.visibility_of_element_located(
            (By.CLASS_NAME, "inventory_list")
        )
    )
    print("1. Đăng nhập thành công")

    # Mở menu bằng JavaScript
    menu = browser.find_element(
        By.ID, "react-burger-menu-btn"
    )
    browser.execute_script("arguments[0].click();", menu)

    # Chờ Logout thực sự hiển thị
    logout = wait.until(
        lambda d: (
            (element := d.find_element(
                By.ID, "logout_sidebar_link"
            )).is_displayed() and element
        )
    )
    print("2. Menu đã mở:", logout.is_displayed())

    # Nhấn Logout
    logout.click()

    # Xác nhận quay lại trang đăng nhập
    wait.until(
        EC.visibility_of_element_located(
            (By.ID, "login-button")
        )
    )

    print("3. URL sau đăng xuất:", browser.current_url)
    print("TC05 PASSED: Đăng xuất thành công.")

except Exception as e:
    print("TC05 FAILED:", type(e).__name__, str(e))
    print("URL:", browser.current_url)

finally:
    browser.quit()
```
